```python
ring_of_fire = ["Bolivia", "Chile", "Ecuador", "Peru", "Costa Rica", "Guatemala", "Mexico", "Japan", "Philippines", "Indonesia", "New Zealand", "Antarctica", "Canada", "Fiji", "Alaska", "Washington", "California","Russia", "Taiwan", "Tonga", "Kermadec Islands" ]
parsed["newringoffire"] = parsed["parsed_place"].isin(ring_of_fire)
parsed["newringoffire"] = (parsed["parsed_place"].isin(ring_of_fire) | parsed["parsed_place"].str.contains("Antarctic"))
```

```python
ringdata = parsed[parsed["newringoffire"] == True]
ringeq = ringdata[ringdata["type"] == "earthquake"]
ringn=len(ringeq)
print(ringn)
nonringdata = parsed[parsed["newringoffire"] == False]
nonringeq = nonringdata[nonringdata["type"] == "earthquake"]
nonringn=len(nonringeq)
print(nonringn)
```
