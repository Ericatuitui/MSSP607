```python
import pandas as pd
df = pd.read_csv('/content/MSSP607/WeeklyModules/Week12/Student Performance Data (1).csv')
df.head()
```

```python
male = df[df['gender'] == 'male']
male.head()
```

```python
df.columns
```

```python
df['test preparation course'].unique()
```

```python
completed = male[male['test preparation course'] == 'completed']
none = male[male['test preparation course'] == 'none']
```

```python
print(len(completed))
print(len(none))
```

```python
score_completed = completed['total score']
score_none = none['total score']
```

```python
print(score_completed.mean())
print(score_none.mean())
```

```python
from scipy.stats import ttest_ind
t_stat, p_value = ttest_ind(score_completed, score_none, equal_var=False)
print("t-statistic:", t_stat)
print("p-value:", p_value)
```

There is a statistically significant difference between the two groups, male students who completed the test preparation course had a higher average total score (212.34) than those who did not (189.13). The difference was statistically significant (t = 6.18, p < 0.05).

Question 3: Do male and female students have significantly different math scores?

```python
df['gender'].unique()
```

```python
malescore = df[df['gender'] == 'male']['math score']
femalescore = df[df['gender'] == 'female']['math score']
```

```python
print(len(malescore))
print(len(femalescore))
```

```python
print(malescore.mean())
print(femalescore.mean())
```

```python
t_stat, p_value = ttest_ind(malescore, femalescore, equal_var=False)
print("t-statistic:", t_stat)
print("p-value:", p_value)
```

I compared the math scores of male and female students using an independent samples t-test. Male students had a higher average math score (68.73) than female students (63.63). The difference was statistically significant (t = 5.40, p < 0.05).
