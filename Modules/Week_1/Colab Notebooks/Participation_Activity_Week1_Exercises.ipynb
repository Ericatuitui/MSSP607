## Exercise 4: Generate the data by running this cell
This will give you a list of numbers to work with in the remaining exercises.

```python
import random
random.seed(0)
salaries = [round(random.random()*1000000, -3) for _ in range(100)]
print(list(salaries))
```

## Exercise 5: Calculating statistics and verifying

Use the data generated above to calulate in code the following statistics without importing the statistics module. Then use the statistics module to verify your results.  Import the statistics module from https://docs.python.org/3/library/statistics.html

### mean

```python
mean=sum(salaries)/len(salaries)
print(mean)
import statistics
print(statistics.mean(salaries))
```

### median

```python
median = (sorted(salaries)[len(salaries)//2 - 1] + sorted(salaries)[len(salaries)//2]) / 2
print(median)
print(statistics.median(salaries))
```

### mode

```python
mode=max(salaries,key=salaries.count)
print(mode)
print(statistics.mode(salaries))
```

### sample variance
Remember to use Bessel's correction.

```python
Variance=sum([(x-mean)**2 for x in salaries])/(len(salaries)-1)
print(Variance)
print(statistics.variance(salaries))
```

### sample standard deviation
Remember to use Bessel's correction.

```python
sd=(sum((x-mean)**2 for x in salaries)/(len(salaries)-1))**0.5
print(sd)
print(statistics.stdev(salaries))
```

## Exercise 6: Calculating more statistics
### range

```python
range=max(salaries)-min(salaries)
print(range)

```

### coefficient of variation

```python
cv=sd/mean
print(cv)
print(statistics.stdev(salaries)/statistics.mean(salaries))
```

### interquartile range

```python
sorted_salaries = sorted(salaries)

Q1 = sorted_salaries[24] * 0.75 + sorted_salaries[25] * 0.25
Q3 = sorted_salaries[74] * 0.25 + sorted_salaries[75] * 0.75

IQR = Q3 - Q1

print(IQR)
print(statistics.quantiles(salaries)[2]-statistics.quantiles(salaries)[0])
```

### quartile coefficent of dispersion

```python
QCQ=(Q3-Q1)/(Q3+Q1)
print(QCQ)
(statistics.quantiles(salaries)[2]-statistics.quantiles(salaries)[0])/(statistics.quantiles(salaries)[2]+statistics.quantiles(salaries)[0])
```

## Exercise 7: Scaling data
### min-max scaling

```python
scaling=[(x-min(salaries))/(max(salaries)-min(salaries)) for x in salaries]
print(min(scaling))
print(max(scaling))
```

### standardizing

```python
z_score = [(x-mean)/sd for x in salaries]
print(z_score)
```

## Exercise 8: Calculating covariance and correlation
### covariance

```python
mean_scaling = sum(scaling) / len(scaling)
mean_z_score = sum(z_score) / len(z_score)

products = []

for x, y in zip(scaling, z_score):
    products.append((x - mean_scaling) * (y - mean_z_score))

covariance = sum(products) / (len(scaling) - 1)

print(covariance)
import numpy as np
print(covariance)
print(np.cov(scaling, z_score))
```

### Pearson correlation coefficient ($\rho$)

```python
sd_scaling = (sum((x-mean_scaling)**2 for x in scaling) / (len(scaling)-1))**0.5

sd_z_score = (sum((y-mean_z_score)**2 for y in z_score) / (len(z_score)-1))**0.5

correlation_coefficient = covariance / (sd_scaling * sd_z_score)

print(correlation_coefficient)
print(np.corrcoef(scaling, z_score)[0, 1])
```
