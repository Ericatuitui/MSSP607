```python
import pandas as pd
df = pd.read_csv('/content/MSSP607/WeeklyModules/Week07/InternetSales.csv', encoding='latin1')
df.head()
```

```python
df['OrderDate'] = pd.to_datetime(df['OrderDate'])
df['Year'] = df['OrderDate'].dt.year
yearly_sales = df.groupby('Year')['SalesAmount'].sum()
yearly_sales
```

```python
import matplotlib.pyplot as plt
plt.bar(yearly_sales.index, yearly_sales.values)
plt.xlabel('Year')
plt.ylabel('Total Sales')
plt.title('Yearly Internet Sales')
plt.xticks(yearly_sales.index)
plt.savefig('Yearly_Internet_Sales.pdf', format='pdf', bbox_inches='tight')
plt.show()
```
