```python
def collatz(number):
  if number % 2 == 0:
    result = number // 2
    print(result, end=" ")
    return result
  elif number % 2 == 1:
    result = 3 * number + 1
    print(result, end=" ")
    return result
number = int(input("Enter an integer "))
while number != 1:
  number = collatz(number)



```
