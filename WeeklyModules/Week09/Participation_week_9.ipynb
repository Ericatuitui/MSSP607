```python
import requests
response = requests.get('https://www.flickr.com/search/?text=cats')
```

```python
response.status_code
```

```python
import bs4

soup = bs4.BeautifulSoup(response.text, 'html.parser')
```

```python
images = soup.select('img')
len(images)
```

```python
str(images[0])
```

```python
for image in images:
    print(image.get('src'))
```

```python
image_url = 'https:' + images[1].get('src')
print(image_url)
```

```python
for image in images[1:]:
    image_url = 'https:' + image.get('src')
    print(image_url)
```

```python
import os

os.makedirs('WeeklyModules/Week09/cats', exist_ok=True)
```

```python
for image in images[1:]:
    image_url = 'https:' + image.get('src')

    res = requests.get(image_url)
    res.raise_for_status()

    image_file = open(os.path.join('WeeklyModules/Week09/cats', os.path.basename(image_url)), 'wb')

    for chunk in res.iter_content(100000):
        image_file.write(chunk)

    image_file.close()
```
