```python
letters = ['a', 'b', 'c', 'd']
letters[int(int('3'*2)//11)]
#'3' is repeated twice to get '33'. After converting it to 33 and dividing by 11, the result is 3, which corresponds to 'd' in the list.
print(letters[-1])
# -1 refers to the last value in dataset which is 'd'
print(letters[:2])
# : means slice, so it selects the values before index 2, which are 'a' and 'b'
```
